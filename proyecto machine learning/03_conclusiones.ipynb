# Informe de solución — Predicción de churn en clientes de telecomunicaciones

La cancelación de clientes (churn) representa uno de los principales retos para las empresas de telecomunicaciones, ya que impacta directamente en los ingresos y la estabilidad del negocio.

En este proyecto se desarrolló un modelo predictivo con el objetivo de **identificar clientes con alta probabilidad de cancelar el servicio**, permitiendo a la empresa implementar acciones preventivas como ofertas personalizadas, promociones o ajustes en los planes contratados.

Para ello, se utilizó información contractual, demográfica y de servicios de los clientes, aplicando técnicas de análisis exploratorio de datos y modelos de clasificación supervisada.


## Contexto del negocio y descripción de los datos

La empresa Interconnect ofrece servicios de telecomunicaciones, incluyendo telefonía fija, servicios de internet y productos adicionales como seguridad en línea, almacenamiento en la nube y streaming.

El objetivo del negocio es reducir la tasa de cancelación de clientes identificando de manera anticipada a aquellos con mayor riesgo de churn, para así implementar estrategias de retención dirigidas.

### Fuentes de datos

El análisis se realizó a partir de cuatro fuentes de datos, relacionadas mediante el identificador único `customerID`:

- **contract.csv**: información del contrato del cliente, incluyendo fechas, tipo de contrato, método de pago y cargos.
- **personal.csv**: datos demográficos del cliente.
- **internet.csv**: información sobre los servicios de internet contratados.
- **phone.csv**: información sobre los servicios telefónicos.

Tras la integración de las fuentes, se obtuvo un conjunto de datos con **7043 clientes** y **21 variables**, que sirvió como base para el análisis exploratorio y el modelado predictivo.


## Análisis Exploratorio de Datos (Hallazgos clave)

El análisis exploratorio de los datos permitió identificar patrones relevantes asociados a la cancelación de clientes:

- **Antigüedad del cliente (tenure):** los clientes con menor tiempo de permanencia presentan una probabilidad significativamente mayor de cancelar el servicio. El churn ocurre con mayor frecuencia durante los primeros meses de relación con la empresa.
- **Tipo de contrato:** los contratos mes a mes muestran una tasa de cancelación considerablemente más alta en comparación con los contratos de uno o dos años, lo que indica que la duración del contrato es un factor clave de retención.
- **Cargos mensuales (MonthlyCharges):** los clientes que cancelan tienden a tener cargos mensuales más elevados, lo que sugiere que el costo del servicio puede influir en la decisión de cancelación.
- **Cargos totales (TotalCharges):** los clientes con churn presentan, en general, cargos totales más bajos, lo cual es consistente con una menor antigüedad.
- **Método de pago:** el uso de *Electronic check* se asocia con una mayor proporción de cancelaciones, mientras que los métodos de pago automáticos presentan mayor retención.
- **Facturación electrónica (PaperlessBilling):** se observa una mayor tasa de churn entre los clientes que utilizan facturación electrónica.

Estos hallazgos permitieron orientar la selección de variables y el enfoque de los modelos predictivos desarrollados posteriormente.


## Desarrollo y evaluación de modelos predictivos

Para abordar el problema de predicción de churn, se entrenaron y evaluaron distintos modelos de clasificación supervisada.  
El desempeño de los modelos se evaluó principalmente mediante la métrica **ROC-AUC**, complementada con **accuracy**, debido al desbalance moderado en la variable objetivo.

### Modelo base

Como punto de referencia, se entrenó un **DummyClassifier**, el cual predice siempre la clase mayoritaria.  
Este modelo obtuvo un **ROC-AUC de 0.5**, confirmando que no posee capacidad predictiva y sirviendo como línea base para comparar modelos reales.

### Regresión Logística

La **Regresión Logística** se utilizó como primer modelo real, debido a su simplicidad e interpretabilidad.  
Este modelo mostró una mejora significativa respecto al modelo base, alcanzando un **ROC-AUC cercano a 0.83**, lo que indica una buena capacidad para distinguir entre clientes que cancelan y los que no.

### Árbol de Decisión

Posteriormente, se entrenó un **Árbol de Decisión** con profundidad limitada para capturar relaciones no lineales y evitar sobreajuste.  
Este modelo superó a la regresión logística, obteniendo un **ROC-AUC cercano a 0.86**, lo que evidencia una mayor capacidad predictiva.

### Random Forest

Finalmente, se entrenó un modelo de **Random Forest**, que combina múltiples árboles de decisión para mejorar la estabilidad y el desempeño.  
Este modelo presentó el mejor resultado entre los evaluados, con un **ROC-AUC cercano a 0.87**, consolidándose como el modelo con mayor capacidad de discriminación para el problema de churn.


## Selección del modelo final

Para la selección del modelo final se comparó el desempeño de los distintos modelos entrenados utilizando principalmente la métrica **ROC-AUC**, ya que el problema presenta un desbalance moderado entre clases y esta métrica evalúa adecuadamente la capacidad de discriminación del modelo.

Los resultados obtenidos fueron los siguientes:

- **DummyClassifier**  
  - ROC-AUC ≈ 0.50  
  - Accuracy ≈ 0.73  

- **Regresión Logística**  
  - ROC-AUC ≈ 0.83  
  - Accuracy ≈ 0.79  

- **Árbol de Decisión**  
  - ROC-AUC ≈ 0.86  
  - Accuracy ≈ 0.81  

- **Random Forest**  
  - ROC-AUC ≈ 0.87  
  - Accuracy ≈ 0.81  

El modelo **Random Forest** presentó el mejor desempeño en términos de **ROC-AUC**, superando consistentemente a los demás modelos evaluados.  
Esta mejora indica una mayor capacidad para distinguir entre clientes que cancelan y los que no, sin una pérdida significativa en exactitud.

Por lo tanto, el Random Forest fue seleccionado como el **modelo final**, al ofrecer el mejor equilibrio entre desempeño predictivo y robustez.


## Conclusiones técnicas

El desarrollo del modelo predictivo permitió demostrar que la cancelación de clientes puede ser anticipada utilizando información contractual, demográfica y de servicios.

El análisis exploratorio evidenció que variables como la **antigüedad del cliente**, el **tipo de contrato**, los **cargos mensuales** y el **método de pago** están fuertemente asociadas con el churn.  
Estos hallazgos fueron coherentes con el comportamiento observado en los modelos predictivos.

En términos de modelado, el **Random Forest** alcanzó un **ROC-AUC cercano a 0.87**, lo que representa una mejora sustancial frente al modelo base y una capacidad sólida de discriminación entre clientes en riesgo y clientes estables.

Estos resultados indican que el modelo es técnicamente adecuado para su uso como herramienta de apoyo en estrategias de retención de clientes, proporcionando predicciones confiables y accionables.


## Recomendaciones de negocio

Con base en los resultados del análisis exploratorio y el desempeño del modelo predictivo, se proponen las siguientes recomendaciones para reducir la cancelación de clientes:

1. **Enfocar estrategias de retención en clientes nuevos**  
   El análisis mostró que los clientes con menor antigüedad presentan una probabilidad significativamente mayor de churn.  
   Se recomienda implementar acciones de retención durante los **primeros 3 a 6 meses**, como descuentos temporales, seguimiento proactivo o beneficios adicionales, especialmente para clientes con cargos mensuales elevados.

2. **Incentivar la migración hacia contratos de mayor duración**  
   Los clientes con contratos mes a mes presentan una tasa de cancelación considerablemente más alta que aquellos con contratos de uno o dos años.  
   Se recomienda ofrecer incentivos para promover contratos de mayor duración, tales como descuentos por permanencia o beneficios exclusivos.

3. **Revisar la estrategia de precios para clientes con cargos mensuales altos**  
   Los clientes que cancelan tienden a tener cargos mensuales más elevados.  
   Se sugiere evaluar esquemas de precios más flexibles, paquetes personalizados o promociones dirigidas a este segmento para reducir la sensibilidad al costo.

4. **Promover métodos de pago automáticos**  
   El método de pago *Electronic check* se asocia con una mayor tasa de churn, mientras que los métodos automáticos muestran mayor retención.  
   Se recomienda incentivar el uso de pagos automáticos mediante beneficios o procesos de afiliación simplificados.

5. **Implementar el modelo predictivo como herramienta preventiva**  
   El modelo Random Forest alcanzó un **ROC-AUC cercano a 0.87**, lo que demuestra una buena capacidad para identificar clientes en riesgo de cancelación.  
   Se recomienda utilizar el modelo para priorizar acciones de retención, enfocándose en clientes con mayor probabilidad estimada de churn antes de que ocurra la cancelación.
